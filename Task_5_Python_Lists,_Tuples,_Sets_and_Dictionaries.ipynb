```python

# 1. Python List

fruits = ["apple", "banana", "mango"]

print("Original list:", fruits)

# Adding an element
fruits.append("orange")
print("After adding:", fruits)

# Removing an element
fruits.remove("banana")
print("After removing:", fruits)

# Searching for an element
print("Is mango present?", "mango" in fruits)

# Updating an element
fruits[0] = "grapes"
print("After updating:", fruits)

# Sorting the list
fruits.sort()
print("After sorting:", fruits)
```

```python

# 2. Python Tuple

numbers = (10, 20, 30, 20, 40)

print("Tuple:", numbers)

# Accessing an element
print("First element:", numbers[0])

# Searching
print("Is 30 present?", 30 in numbers)

# Counting occurrences
print("Count of 20:", numbers.count(20))

# Finding the index
print("Index of 30:", numbers.index(30))

# Length
print("Length:", len(numbers))
```

```python

# 3. Python Set

numbers = {10, 20, 30, 20, 40}

print("Original set:", numbers)

# Adding an element
numbers.add(50)
print("After adding:", numbers)

# Removing an element
numbers.remove(20)
print("After removing:", numbers)

# Searching
print("Is 30 present?", 30 in numbers)

# Adding a duplicate
numbers.add(30)
print("After adding duplicate:", numbers)

# Set operations
a = {1, 2, 3}
b = {3, 4, 5}

print("Union:", a.union(b))
print("Intersection:", a.intersection(b))
```

```python

# 4. Python Dictionary

student = {
    "name": "Sujitha",
    "roll_no": 101,
    "branch": "CSD",
    "marks": 85
}

print("Student details:", student)

# Accessing a value
print("Name:", student["name"])

# Adding a new key-value pair
student["city"] = "Guntur"
print("After adding:", student)

# Updating a value
student["marks"] = 90
print("After updating:", student)

# Removing a key-value pair
student.pop("branch")
print("After removing:", student)

# Searching for a key
print("Is roll_no present?", "roll_no" in student)

# Displaying all keys and values
print("Keys:", student.keys())
print("Values:", student.values())
```

```python

# Student Dataset

students = [
    {"name": "Ravi", "marks": 85},
    {"name": "Sita", "marks": 92},
    {"name": "Ram", "marks": 76},
    {"name": "Priya", "marks": 88}
]

# Display all students
for student in students:
    print(student)

# Find students scoring 85 or more
print("\nStudents scoring 85 or more:")

for student in students:
    if student["marks"] >= 85:
        print(student["name"], student["marks"])

# Calculate average marks
total = sum(student["marks"] for student in students)
average = total / len(students)

print("Average marks:", average)
```
